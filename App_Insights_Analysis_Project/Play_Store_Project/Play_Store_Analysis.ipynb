An executed walkthrough of the 25 assignment questions. Run from this project folder. The supplied data describes a historical snapshot, not today’s market.

## B01 What is the average app rating?
The mean rating is 4.173 out of 5 across 8,196 rated apps. Each app has equal weight. The 1,463 apps without a rating are excluded from this mean, but retained for counts, installs and other relevant calculations.

Method: Mean of non-missing Rating values.

```python
print(d.Rating.mean(), d.Rating.count())
```
